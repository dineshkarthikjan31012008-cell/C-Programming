```python
#boolean algebra!!!!!!!!!!!!!
import pandas as pd
a = pd.read_csv("/content/sensor.csv")
a.loc[a["Humidity"]<50]

a.loc[a["Temperature"]>25]

print(a.loc[(a["Temperature"]>25) & (a["Humidity"]<50)])

print(a.loc[(a["Temperature"]>25) | (a["Humidity"]<50)])

print(a.loc[~(a["Temperature"]>25) & (a["Humidity"]<50)])



```
